# Four and six order parameters

Compare both closures with the microscopic simulation on one sampled projector. The compact model evolves signal and noise contractions. The six-state model evolves contractions against P, C, and C². Here C is the signal projector restricted to the pretrained subspace.

```python
from experiments.closure.compare_degree_two import DEFAULTS, QUICK, compute, plot
import matplotlib.pyplot as plt
import numpy as np

settings = DEFAULTS | QUICK | {"alpha": 0.5, "sft_epochs": 8, "rl_epochs": 3, "quick": True}
data = compute(settings)
plot(data, settings)
plt.show()
```

```python
# Match J_signal and Q_signal at initialization in the two coordinate systems.
np.testing.assert_allclose(data["six"][0, [1, 4]], data["four"][0, [0, 2]], atol=1e-12)
print("Terminal six-state vector (J0, J1, J2, Q0, Q1, Q2)")
print(data["six"][-1])
```
