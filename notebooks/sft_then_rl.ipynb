# SFT followed by RL

Vary the SFT warm start, episode length, reward threshold, or learning-rate policy. Each epoch comprises one microscopic update per ambient dimension. The vertical line marks the change of phase.

```python
import matplotlib.pyplot as plt
import numpy as np
from multistage_pipeline.pretraining.projector import population_geometry, population_moments
from multistage_pipeline.dynamics.pipeline import Dynamics, trajectory, accuracy

alpha, signal_fraction = 1.6, 0.2
cfg = Dynamics(dimension=100, episode_length=6, epsilon=0.2, threshold=6)
geometry = population_geometry(population_moments(alpha, signal_fraction, cfg.delta, quick=True))
sft_epochs, rl_epochs = 5, 25
result = trajectory(cfg, geometry, sft_epochs, rl_epochs)
epochs = result.steps / cfg.dimension
```

```python
fig, (left, right) = plt.subplots(1, 2, figsize=(10, 3))
left.plot(epochs, accuracy(result.states, cfg.delta))
left.set(ylabel="Clean accuracy", xlabel="Completed epochs")
right.plot(epochs, result.states)
right.set(xlabel="Completed epochs", ylabel="Order parameters")
right.legend(["J signal", "J noise", "Q signal", "Q noise"])
for ax in (left, right):
    ax.axvline(result.switch_step / cfg.dimension, color="gray", linestyle=":")
plt.show()
```

Use zero SFT epochs for pure RL or zero RL epochs for pure SFT. For the plateau comparison, the experiment script aligns each trajectory at its first RL update.
